# Drowning temporal Kaggle worker
Private GPU notebook. Default mode is `prepare_only`; this notebook never silently trains.
`remote_smoke` runs a synthetic forward/backward contract check only. `train_candidate` requires approved image crops, reviewed video tracks and a verified split manifest.


```python
# Generated from the current repository source. Do not edit the embedded bundle by hand.
import base64, hashlib, io, os, sys, zipfile, importlib.util
from pathlib import Path

SOURCE_BUNDLE_B64 = ''
SOURCE_BUNDLE_SHA256 = hashlib.sha256(base64.b64decode(SOURCE_BUNDLE_B64)).hexdigest()
WORK_DIR = Path("/kaggle/working/drowning-project")
WORK_DIR.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(base64.b64decode(SOURCE_BUNDLE_B64))) as archive:
    archive.extractall(WORK_DIR)
sys.path.insert(0, str(WORK_DIR / "src"))
RUN_MODE = os.environ.get("DROWN_RUN_MODE", 'prepare_only')
assert RUN_MODE in {"prepare_only", "remote_smoke", "train_candidate"}, RUN_MODE
DEPENDENCIES = {"PyYAML": importlib.util.find_spec("yaml") is not None,
                "Pillow": importlib.util.find_spec("PIL") is not None,
                "defusedxml": importlib.util.find_spec("defusedxml") is not None,
                "torch": importlib.util.find_spec("torch") is not None,
                "torchvision": importlib.util.find_spec("torchvision") is not None,
                "cv2": importlib.util.find_spec("cv2") is not None}
print({"run_mode": RUN_MODE, "default_mode": "prepare_only", "private_kernel": True,
       "dependencies": DEPENDENCIES})
```

```python
from pathlib import Path
import os, json
core_missing = [name for name in ("PyYAML", "Pillow", "defusedxml") if not DEPENDENCIES[name]]
input_root = os.environ.get("DROWN_DATASET_ROOT", "")
temporal_path = os.environ.get("DROWN_TEMPORAL_CANONICAL_JSONL", "")
split_path = os.environ.get("DROWN_SPLIT_MANIFEST", "")
TEMPORAL_WINDOW_REPORT = None
report = {"dependency_status": DEPENDENCIES, "missing_core_dependencies": core_missing,
          "dataset_root_configured": bool(input_root),
          "dataset_root_exists": bool(input_root and Path(input_root).is_dir()),
          "image_crop_manifest_exists": bool(os.environ.get("DROWN_IMAGE_CROP_MANIFEST") and Path(os.environ["DROWN_IMAGE_CROP_MANIFEST"]).is_file()),
          "mapping_review_status": "unknown", "dataset_scan": None,
          "temporal_preflight": {"status": "blocked", "reason": "No reviewed video/person/time source configured"},
          "training_started": False}
if not core_missing:
    from drowning.data.registry import load_source
    from drowning.data.adapters import scan_source
    from drowning.data.audit import audit_jsonl
    from drowning.data.windows import build_temporal_windows
    import yaml
    source_config = WORK_DIR / "configs/datasets/alanoud_swimming_drowning_v4.yaml"
    source = load_source(source_config)
    source["paths"]["root_env"] = "DROWN_DATASET_ROOT"
    report["mapping_review_status"] = source.get("mapping_review", {}).get("status", "unknown")
    if report["dataset_root_exists"]:
        report["dataset_scan"] = scan_source(source, Path(input_root))
    if temporal_path and Path(temporal_path).is_file() and split_path and Path(split_path).is_file():
        temporal_audit = audit_jsonl([Path(temporal_path)])
        report["temporal_audit"] = {"valid": temporal_audit["valid"],
                                    "issue_count": len(temporal_audit["issues"]),
                                    "same_box_label_conflicts": len(temporal_audit["same_box_label_conflicts"])}
        recipe = yaml.safe_load((WORK_DIR / "configs/recipes/temporal.yaml").read_text(encoding="utf-8"))
        sampling = recipe["sampling"]
        split_assignments = {}
        for line in Path(split_path).read_text(encoding="utf-8").splitlines():
            if line.strip():
                entry = json.loads(line)
                split_assignments[str(entry["media_id"])] = {
                    "split": str(entry["split"]),
                    "global_group_id": str(entry.get("global_group_id", ""))}
        window_report = build_temporal_windows(
            Path(temporal_path), sequence_length=int(sampling["sequence_length"]),
            sample_period_ms=int(sampling["sample_period_ms"]),
            max_track_gap_ms=int(sampling["max_track_gap_ms"]),
            min_valid_ratio=float(sampling["min_valid_ratio"]),
            split_assignments=split_assignments)
        TEMPORAL_WINDOW_REPORT = window_report
        report["temporal_preflight"] = {key: value for key, value in window_report.items() if key != "windows"}
        if not temporal_audit["valid"] or temporal_audit["same_box_label_conflicts"]:
            report["temporal_preflight"]["status"] = "blocked"
            report["temporal_preflight"].setdefault("blockers", []).append("canonical_temporal_audit_failed_or_conflicted")
elif not core_missing:
    report["temporal_preflight"] = {"status": "blocked", "reason": "Provide temporal canonical JSONL and verified split manifest"}
print(json.dumps(report, ensure_ascii=False, indent=2))
```

```python
if RUN_MODE == "remote_smoke":
    import torch
    from drowning.ml.model import DrowningTemporalModel
    model = DrowningTemporalModel(hidden_size=16).to("cuda" if torch.cuda.is_available() else "cpu")
    device = next(model.parameters()).device
    frames = torch.randn(1, 4, 3, 64, 64, device=device)
    output = model(frames)
    loss = output["image_logits"].mean() + output["temporal_logits"].mean()
    loss.backward()
    print({"smoke": "passed", "device": str(device), "image_shape": list(output["image_logits"].shape),
           "temporal_shape": list(output["temporal_logits"].shape), "model_candidate": False})
else:
    print("prepare_only complete; this mode does not train or create a model candidate")
```

```python
if RUN_MODE == "train_candidate":
    import hashlib, os, json, torch, yaml
    from torch.utils.data import DataLoader
    from drowning.ml.datasets import CanonicalImageCropDataset, TemporalVideoWindowDataset
    from drowning.ml.model import DrowningTemporalModel
    from drowning.ml.training import train_stage_a_epoch, train_stage_b_epoch, evaluate_image_model, evaluate_temporal_model
    from drowning.ml.bundle import save_bundle

    if report["missing_core_dependencies"]:
        raise RuntimeError("Candidate training blocked by missing core dependencies")
    if not torch.cuda.is_available():
        raise RuntimeError("Candidate training requires the selected Kaggle GPU runtime; no CPU fallback is allowed")
    if report["temporal_preflight"].get("status") != "ready_for_review" or not report["temporal_preflight"].get("source_video_disjointness_verified"):
        raise RuntimeError("Candidate training blocked: reviewed temporal annotations and verified split groups are required")
    if not report.get("temporal_audit", {}).get("valid") or report["temporal_audit"].get("same_box_label_conflicts", 0):
        raise RuntimeError("Candidate training blocked: canonical temporal audit failed or contains label conflicts")
    if TEMPORAL_WINDOW_REPORT is None or not TEMPORAL_WINDOW_REPORT.get("windows"):
        raise RuntimeError("Candidate training blocked: no causal windows passed the verified split preflight")

    required_inputs = {"DROWN_IMAGE_CROP_MANIFEST", "DROWN_TEMPORAL_CANONICAL_JSONL",
                       "DROWN_VIDEO_ROOT", "DROWN_SPLIT_MANIFEST"}
    missing_inputs = sorted(name for name in required_inputs if not os.environ.get(name))
    if missing_inputs:
        raise RuntimeError("Candidate training inputs are missing: " + ", ".join(missing_inputs))
    image_cfg = yaml.safe_load((WORK_DIR / "configs/recipes/image.yaml").read_text(encoding="utf-8"))
    temporal_cfg = yaml.safe_load((WORK_DIR / "configs/recipes/temporal.yaml").read_text(encoding="utf-8"))
    input_cfg = image_cfg["input"]
    size = int(input_cfg["size"][0])
    preprocessing = {"size": input_cfg["size"], "context_fraction": input_cfg["crop_context_fraction"],
                     "mean": input_cfg["mean"], "std": input_cfg["std"]}
    image_train = CanonicalImageCropDataset(Path(os.environ["DROWN_IMAGE_CROP_MANIFEST"]), split="train",
                                            output_size=size, mean=input_cfg["mean"], std=input_cfg["std"])
    image_validation = CanonicalImageCropDataset(Path(os.environ["DROWN_IMAGE_CROP_MANIFEST"]), split="validation",
                                                 output_size=size, mean=input_cfg["mean"], std=input_cfg["std"])
    verified_window_manifest = Path("/kaggle/working/verified-window-manifest.jsonl")
    verified_window_manifest.write_text("".join(json.dumps(window, ensure_ascii=False) + "\n"
                                                       for window in TEMPORAL_WINDOW_REPORT["windows"]), encoding="utf-8")
    window_train = TemporalVideoWindowDataset(Path(os.environ["DROWN_TEMPORAL_CANONICAL_JSONL"]),
                                              verified_window_manifest,
                                              video_root=Path(os.environ["DROWN_VIDEO_ROOT"]), split="train",
                                              output_size=size, context_fraction=input_cfg["crop_context_fraction"],
                                              mean=input_cfg["mean"], std=input_cfg["std"])
    window_validation = TemporalVideoWindowDataset(Path(os.environ["DROWN_TEMPORAL_CANONICAL_JSONL"]),
                                                   verified_window_manifest,
                                                   video_root=Path(os.environ["DROWN_VIDEO_ROOT"]), split="validation",
                                                   output_size=size, context_fraction=input_cfg["crop_context_fraction"],
                                                   mean=input_cfg["mean"], std=input_cfg["std"])
    def require_three_classes(rows, name, label_key):
        present = {row[label_key] for row in rows if row.get("training_eligible") is True and row.get(label_key)}
        if present != {"drowning", "swimming", "out_of_water"}:
            raise RuntimeError(f"{name} does not cover all three reviewed classes: {sorted(present)}")
    require_three_classes(image_train.rows, "Image train split", "canonical_label")
    require_three_classes(image_validation.rows, "Image validation split", "canonical_label")
    require_three_classes(window_train.windows, "Temporal train split", "label")
    require_three_classes(window_validation.windows, "Temporal validation split", "label")

    device = "cuda"
    model = DrowningTemporalModel(hidden_size=int(temporal_cfg["model"]["hidden_size"]),
                                  pretrained_encoder=os.environ.get("DROWN_USE_PRETRAINED", "false").lower() == "true").to(device)
    image_loader = DataLoader(image_train, batch_size=int(image_cfg["training"]["batch_size"]), shuffle=True, num_workers=0)
    image_validation_loader = DataLoader(image_validation, batch_size=int(image_cfg["training"]["batch_size"]), shuffle=False, num_workers=0)
    temporal_loader = DataLoader(window_train, batch_size=int(temporal_cfg["training"]["batch_size"]), shuffle=True, num_workers=0)
    temporal_validation_loader = DataLoader(window_validation, batch_size=int(temporal_cfg["training"]["batch_size"]), shuffle=False, num_workers=0)
    image_epochs = int(os.environ.get("DROWN_IMAGE_EPOCHS", "5"))
    temporal_epochs = int(os.environ.get("DROWN_TEMPORAL_EPOCHS", "5"))
    image_optimizer = torch.optim.AdamW(model.parameters(), lr=float(image_cfg["training"]["learning_rate"]))
    image_history = [train_stage_a_epoch(model, image_loader, image_optimizer, device=device) for _ in range(image_epochs)]
    image_metrics = evaluate_image_model(model, image_validation_loader, device=device)
    temporal_optimizer = torch.optim.AdamW(model.parameters(), lr=float(temporal_cfg["training"]["learning_rate"]))
    temporal_history = [train_stage_b_epoch(model, temporal_loader, temporal_optimizer, device=device,
                                             image_aux_weight=float(temporal_cfg["training"]["image_auxiliary_loss_weight"]))
                        for _ in range(temporal_epochs)]
    temporal_metrics = evaluate_temporal_model(model, temporal_validation_loader, device=device)
    window_train.close()
    window_validation.close()
    fingerprint = hashlib.sha256()
    for env_name in sorted(required_inputs):
        path = Path(os.environ[env_name])
        if path.is_file():
            with path.open("rb") as stream:
                for chunk in iter(lambda: stream.read(1024 * 1024), b""):
                    fingerprint.update(chunk)
    dataset_fingerprint = fingerprint.hexdigest()
    recipe_fingerprint = hashlib.sha256(
        (WORK_DIR / "configs/recipes/image.yaml").read_bytes()
        + (WORK_DIR / "configs/recipes/temporal.yaml").read_bytes()).hexdigest()
    candidate_id = __import__("uuid").uuid4().hex
    output_dir = Path("/kaggle/working/drowning-candidate")
    output_dir.mkdir(parents=True, exist_ok=False)
    save_bundle(model, output_dir / "model-bundle", completion_kind="candidate", temporal_trained=True,
                preprocessing=preprocessing, dataset_fingerprint=dataset_fingerprint,
                recipe_fingerprint=recipe_fingerprint, code_fingerprint=SOURCE_BUNDLE_SHA256,
                training_job_id=os.environ.get("DROWN_TRAINING_JOB_ID"),
                remote_kernel_ref=os.environ.get("KAGGLE_KERNEL_ID"),
                remote_kernel_version=os.environ.get("KAGGLE_KERNEL_VERSION"),
                model_version_id=candidate_id,
                temporal_contract={"sequence_length": int(temporal_cfg["sampling"]["sequence_length"]),
                                   "sample_period_ms": int(temporal_cfg["sampling"]["sample_period_ms"]),
                                   "max_track_gap_ms": int(temporal_cfg["sampling"]["max_track_gap_ms"]), "causal": True})
    metrics = {"completion_kind": "candidate", "approved": False, "device": torch.cuda.get_device_name(0),
               "candidate_id": candidate_id,
               "dataset_fingerprint": dataset_fingerprint, "image_stage_history": image_history,
               "temporal_stage_history": temporal_history, "image_validation": image_metrics,
               "temporal_validation": temporal_metrics, "test_split_used": False}
    (output_dir / "metrics.json").write_text(json.dumps(metrics, ensure_ascii=False, indent=2), encoding="utf-8")
    candidate_archive = Path("/kaggle/working/candidate-output.zip")
    with zipfile.ZipFile(candidate_archive, "w", compression=zipfile.ZIP_DEFLATED) as archive:
        for file in sorted(output_dir.rglob("*")):
            if file.is_file():
                archive.write(file, file.relative_to(output_dir).as_posix())
    print(json.dumps({"candidate_bundle": str(output_dir / "model-bundle"), "metrics": str(output_dir / "metrics.json"),
                      "candidate_archive": str(candidate_archive), "approved": False, "model_available": False}, ensure_ascii=False, indent=2))
```
